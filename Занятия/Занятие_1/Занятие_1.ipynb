### Условия (if), циклы (for, while), функции (def)

```python
x = 10

if x == 11: # условие
  print("x = 11")
elif x ==9: # или + условие
  print("x == 9")
else: # или без условия
  print("x != 11 и x != 9")
```

```python
# Составные условия

x = 5

if x > 10 and x < 20: # логическое И
  print("x больше 10 и меньше 20")

if x < 10 or x > 20: # логическое ИЛИ
  print("x меньше 10 или больше 20")

if not (x > 10 and x < 20): # логическое НЕ
  print("x не больше 10 или не меньше 20")
```

```python
molecules = ["H2O", "CO2", "NH3", "CH4"]

for molecule in molecules:
    print("Молекула: " + molecule)
```

```python
# Генерация первых 20 чисел ряда Фибоначи

number_1, number_2 = 1, 1

print(number_1)
print(number_2)

summ = number_1 + number_2

for i in range(2, 20):
    next_number = number_1 + number_2
    print(next_number)
    number_1 = number_2
    number_2 = next_number
    summ = summ + next_number

print('Сумма первых 10 чисел фибоначи', summ)
```

```python
# Вложенные циклы
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
```

```python
# Цикл while
x = 0
while x <= 6:
  x += 1
  print(x)
```

```python
# Поиск с остановкой 
x = 0
while x <= 20:
    if x < 10:
        x += 1
    else:
        break 
    print(x)
```

```python
# Пропуск итерации
x = 0
while x <= 20:
    x += 1
    if x % 2 == 0: # если число четное
        continue 
    print(x)
```

```python
# Функции
def function(x):
    return x * 2

print(function(5)) # 10
```

```python
def molecular_weight(atoms):
    # Вычисляет молярную массу молекулы

    weights = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999, "S": 32.06}
    return sum(weights[el] * count for el, count in atoms.items())

print(molecular_weight({"H": 2, "O": 2}))  # 18.015
```

## 2. ☕︎ Объектно-ориентированное программирование (ООП)

```python
class Atom:
    def __init__(self, symbol: str, x: float, y: float, z: float):
        self.symbol = symbol
        self.x = x
        self.y = y
        self.z = z

    def distance_to(self, other: "Atom") -> float:
        return ((self.x - other.x)**2 + (self.y - other.y)**2 + (self.z - other.z)**2) ** 0.5

    def __repr__(self):
        return f"Atom({self.symbol}, {self.x}, {self.y}, {self.z})"


class Molecule:
    def __init__(self, name: str, formula: str):
        self.name = name
        self.formula = formula
        self.atoms: list[Atom] = []

    def add_atom(self, atom: Atom):
        self.atoms.append(atom)

    def __repr__(self):
        return f"Molecule({self.name}, {len(self.atoms)} atoms)"
```
