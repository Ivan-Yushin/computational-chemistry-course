````python
def molecular_weight(atoms):
    # Вычисляет молярную массу молекулы

    weights = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999, "S": 32.06}
    return sum(weights[el] * count for el, count in atoms.items())

print(molecular_weight({"H": 2, "O": 2}))  # 34.014
````

````python
# Использование аргументов по умолчанию
def hartree_to_kj_mol(energy, digits=2):   # digits можно не передавать
    return round(energy * 2625.5, digits)

print(hartree_to_kj_mol(-76.4))            # 2 знака после запятой по умолчанию
print(hartree_to_kj_mol(-76.4, digits=6))  # вызов по имени
print(hartree_to_kj_mol(digits=0, energy=-76.4))  # порядок не важен
````

````python
# Возврат нескольких значений
def bond_stats(lengths):
    return min(lengths), max(lengths), sum(lengths) / len(lengths)

shortest, longest, average = bond_stats([1.09, 1.10, 1.08])

print(f"Shortest: {shortest}, Longest: {longest}, Average: {average}")

print("shortest: " + str(shortest))
print("longest: " + str(longest))
print("average: " + str(average))
````

````python
# Рекурсивные функции
def fib(n):
    """n-е число Фибоначчи (нумерация с 1: 1, 1, 2, 3, 5, ...)."""
    if n <= 2:                       # базовый случай: F(1) = F(2) = 1
        return 1
    return fib(n - 1) + fib(n - 2)   # рекурсивный шаг: функция зовёт саму себя


print(fib(1))    # 1
print(fib(10))   # 55

# Те же 20 чисел, что и в цикле выше
for i in range(1, 21):
    print(fib(i), end=" ")
````

### 1.5 Классы

````python
# Описание молекулы
water = {
    "name": "water",
    "atoms": {"H": 2, "O": 1},
}

def molar_mass(molecule):
    # Вычисляет молярную массу молекулы
    weights = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999, "S": 32.06}
    weight = 0
    for el, count in molecule["atoms"].items():
        weight += weights[el] * count
    return weight

water["mass"] = molar_mass(water)
print(water)
````

````python
class Molecule: 
    def __init__(self, name, atoms):
        self.name = name
        self.atoms = atoms
        self.mass = self.molar_mass()

    def molar_mass(self):
        weights = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999, "S": 32.06}# "Na": 22.990, "Cl": 35.453
        weight = 0
        for el, count in self.atoms.items():
            weight += weights[el] * count
        return weight

water = Molecule("вода", {"H": 2, "O": 1})

print(water)
print(water.name)
print(water.atoms)
print(water.mass)
````

````python
# Объекты можно складывать в список, как любые другие значения

water = Molecule("Хлорид натрия", {"Na": 1, "Cl": 1})
ammonia = Molecule("Аммиак", {"N": 1, "H": 3})
glucose = Molecule("Глюкоза", {"C": 6, "H": 12, "O": 6})

substances = [water, ammonia, glucose]
for m in substances:
    print(m.name, ":", round(m.mass, 2))
````

### Задание 3 (Hard)

Слабая одноосновная кислота в водном растворе диссоциирует неполностью:

$$\mathrm{HA} \rightleftarrows \mathrm{H^+} + \mathrm{A^-}, \qquad K_a = \frac{[\mathrm{H^+}][\mathrm{A^-}]}{[\mathrm{HA}]}$$

Если обозначить $x = [\mathrm{H^+}]$, то при исходной концентрации кислоты $C_0$ равновесие описывается уравнением

$$\frac{x^2}{C_0 - x} = K_a \quad \Longleftrightarrow \quad f(x) = x^2 + K_a x - K_a C_0 = 0$$

В учебниках обычно пользуются приближением $C_0 - x \approx C_0$, откуда $[\mathrm{H^+}] \approx \sqrt{K_a C_0}$. Это допустимо, только если степень диссоциации мала:

$$\alpha = \frac{x}{C_0} \cdot 100\% < 5\%$$

Точный корень уравнения находят численно — **методом половинного деления (бисекции)**. Метод основан на том, что если на концах отрезка функция имеет разные знаки, то внутри него есть корень:

```text
Отрезок [a, b],  f(a) < 0,  f(b) > 0
        ┌──────────────┬──────────────┐
        a              m              b        m = (a + b) / 2
        
  если f(a) и f(m) разных знаков  ->  корень слева,  b = m
  иначе                           ->  корень справа, a = m
  
  повторять, пока  b - a > eps
```

Для нашего уравнения подходит отрезок $[0,\ C_0]$: на нём $f(0) = -K_a C_0 < 0$, а $f(C_0) = C_0^2 > 0$.

Напишите программу, которая:
1. Реализует функцию $f(x)$ для заданных $K_a$ и $C_0$.
2. Находит корень уравнения методом половинного деления с точностью `eps = 1e-12`, подсчитывая число выполненных итераций.
3. Рассчитывает pH точного решения по формуле $\mathrm{pH} = -\lg[\mathrm{H^+}]$.
4. Рассчитывает pH по приближённой формуле $[\mathrm{H^+}] = \sqrt{K_a C_0}$ и находит расхождение двух способов.
5. Рассчитывает степень диссоциации $\alpha$ и сообщает, допустимо ли для данной кислоты приближение (критерий 5%).

P.S.: Корень уравнения должен быть найден **методом половинного деления**, а не по формуле корней квадратного уравнения. Число шагов заранее неизвестно — используйте цикл `while`.

Для расчёта pH используйте функцию десятичного логарифма из модуля math:
```python
import math
C_H = 1e-5
pH = -math.log10(C_H)
```

  Расхождение: 0.003 ед. pH
