## 🚀 Задание 1

Модель идеального газа описывается уравнением Менделеева-Клапейрона: 

$$P_{\text{ideal}} V = n R T, \qquad n = \frac{m}{M}$$

где:
* $R = 8.314 \text{ Дж/(моль}\cdot\text{К)}$ — универсальная газовая постоянная;

Однако модель не учитывает собственный объём молекул и межмолекулярное притяжение. Для более точных расчётов реальных газов используется уравнение Ван-дер-Ваальса:

$$P_{\text{VdW}} = \frac{n R T}{V - n b} - a \frac{n^2}{V^2}$$

где:
* $a$ — поправка на межмолекулярное притяжение ($\text{Па}\cdot\text{м}^6/\text{моль}^2$);
* $b$ — поправка на собственный объём молекул ($\text{м}^3/\text{моль}$).

Напишите программу, которая берет известные значения массы $m (г)$, молярной массы $M (г/моль)$, объёма $V (м^3)$, температуры $T (K)$ и констант $a$ и $b$ из заданных переменных и рассчитывает:
1. Давление по модели идеального газа ($P_{\text{ideal}}$) в атмосферах.
2. Давление по модели Ван-дер-Ваальса ($P_{\text{VdW}}$) в атмосферах ($1 \text{ атм} = 101325 \text{ Па}$).
3. Относительное отклонение результатов в процентах:

$$\Delta = \frac{|P_{\text{VdW}} - P_{\text{ideal}}|}{P_{\text{ideal}}} \times 100\%$$

4. Если отклонение составляет больше 10%, программа сообщает об этом пользователю
