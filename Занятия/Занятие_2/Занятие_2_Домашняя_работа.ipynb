## 🐈 Задание 1

В простейшей модели Хюккеля π-электроны сопряжённой молекулы описываются
матрицей, построенной по правилу: на диагонали матрицы расположены кулоновские
интегралы α, а между соседними атомами углерода - резонансные интегралы β.
Всё остальное равно нулю. Если принять также α=2β, для бутадиена CH2=CH–CH=CH2 матрица Хюккеля приобретает следующий вид:

$$H = \begin{pmatrix} 2&1&0&0 \\ 1&2&1&0 \\ 0&1&2&1 \\ 0&0&1&2 \end{pmatrix}$$

Вам нужно: 
* Написать функцию, которая по числу атомов `n` строит матрицу Хюккеля
  линейного полиена. (Лучше для решения использовать функцию np.eye, передав ей параметр k, который сдвигает диагональ из единиц)
* Найти энергии молекулярных орбиталей (собственные значения матрицы Хюккеля) для молекулы `n = 7`. 
* Выписать собственные векторы. Убедиться, что каждый из них имеет единичную
  длину, а любые два взаимно ортогональны.
* Найти разницу энергий между верхней занятой и нижней свободной орбиталью
  (HOMO-LUMO gap) для цепей разной длины `n = 3-9`. Как она меняется с ростом цепи и как это может отразиться на спектре поглощения вещества?

## 📈 Задание 3

Студент-лаборант и научный руководитель готовят статью, но редактор просит
прислать рисунок в более высоком качестве и в форматах png и tiff.
Оригинальный скрипт, которым строился график, утрачен. Сохранились только
сам рисунок `enzyme_reference.png` и исходные данные `enzyme_kinetics.csv`.

В файле три столбца: концентрация субстрата в мМ, измеренная скорость реакции
в мкмоль/(мин·мг) и погрешность измерения скорости.

Уравнение Михаэлиса–Ментен и его линеаризация по Лайнуиверу–Бэрку:

$$v = \frac{V_{max}[S]}{K_m + [S]} \qquad\Longrightarrow\qquad
\frac{1}{v} = \frac{K_m}{V_{max}}\cdot\frac{1}{[S]} + \frac{1}{V_{max}}$$

Пункты задания:
* Найти $V_{max}$ и $K_m$ из линеаризации Лайнуивера-Бэрка.
* Воспроизвести рисунок максимально близко к образцу: две панели, точки
  с усами погрешностей (можно использовать v_err), кривая подгонки, вспомогательные линии и подписи.
* Сохранить результат в форматах png и tiff с качеством `dpi=600`.
